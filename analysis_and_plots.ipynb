```python
# Cell 2: Load metrics from a completed benchmark run
import json
from pathlib import Path


def load_benchmark_metrics(run_dir: str):
    """Load metrics.json from a BenchmarkRunner output directory.

    Args:
        run_dir: path to a benchmark run directory, e.g.
                 "benchmark_results/bench_20260520_234249_*/"

    Returns:
        dict with planner_name -> aggregated metrics
    """
    path = Path(run_dir) / 'metrics.json'
    if not path.exists():
        print(f'  Not found: {path}')
        return {}
    with open(path) as f:
        data = json.load(f)
    results = data.get('results', {})
    for pname, m in results.items():
        print(f'  {pname}: SR={m["success_rate"]:.2%}  '
              f'PL={m["avg_path_length"]:.2f}  '
              f'TC={m["avg_time_cost"]:.4f}  '
              f'IT={m["avg_inference_time_ms"]:.1f}ms')
    return results


# Example usage (uncomment and set path to use):
# load_benchmark_metrics('benchmark_results/bench_20260520_234249_*/')
print('[OK] load_benchmark_metrics() defined - ready to use')
```
